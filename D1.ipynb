```python
import pandas as pd
from scipy import stats

url = "https://raw.githubusercontent.com/nikhil7280/student-performance-multiple-linear-regression/main/student_performance.csv"
df = pd.read_csv(url)

pearson_corr, pearson_p = stats.pearsonr(df['Hours Studied'], df['Performance Index'])
spearman_corr, spearman_p = stats.spearmanr(df['Hours Studied'], df['Performance Index'])

print(f"Pearson Correlation: r = {pearson_corr:.4f}, p-value = {pearson_p}")
print(f"Spearman Correlation: r = {spearman_corr:.4f}, p-value = {spearman_p}")

contingency_table = pd.crosstab(df['Extracurricular Activities'], df['Performance Index'] > 50)
chi2, p_val, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2:.4f}, p-value = {p_val}")

population_mean = 55.22
t_stat, t_p_value = stats.ttest_1samp(df['Performance Index'], population_mean)

print(f"T-statistic: {t_stat:.4f}, p-value = {t_p_value}")
```
